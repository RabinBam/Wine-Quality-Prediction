```python
#outlier finding function using Inter Quartile range 

def outlierFinder(dataframe: pd.DataFrame):
    features = dataframe.select_dtypes(include="number").columns
    outlierCounting = {} 
    for col in features:
        Q1 = dataframe[col].quantile(0.25)
        Q3 = dataframe[col].quantile(0.75)
        IQR = Q3-Q1 #Interquartile range value
        
        lowerbound = Q1 - 1.5 * IQR
        upperbound = Q3 - 1.5 * IQR 

        Outliers = dataframe[
            (dataframe[col]<lowerbound) | (first_data[col] > upperbound)
        ]

        outlierCounting[col] = len(Outliers)
    return outlierCounting
    
```

```python
outliers1 = outlierFinder(first_data)
outliers2 = outlierFinder(second_data)
```
