```python
#outlier finding function using Inter Quartile range 

def outlierFinder(dataframe: pd.DataFrame):
    features = dataframe.select_dtypes(include="number").columns
    outlierCounting = {} 
    for col in features:
        Q1 = dataframe[col].quantile(0.25)
        Q3 = dataframe[col].quantile(0.75)
        IQR = Q3-Q1 #Interquartile range value
        
        lowerbound = Q1 - 1.5 * IQR
        upperbound = Q3 - 1.5 * IQR 

        Outliers = dataframe[
            (dataframe[col]<lowerbound) | (first_data[col] > upperbound)
        ]

        outlierCounting[col] = len(Outliers)
    return outlierCounting
    
```

```python
outliers1 = outlierFinder(first_data)
outliers2 = outlierFinder(second_data)
```

```python
#combining the data
first_data["wine_type"] = 0
second_data["wine_type"] = 1

data = pd.concat([first_data, second_data], ignore_index=True)

print("Combined shape:", data.shape)
display(data.head())
```

### feature analysis 

```python
plt.figure(figsize=(8, 5))
for quality_class in sorted(data["quality"].unique()):
    subset = data[data["quality"] == quality_class]
    plt.scatter(
        subset["volatile acidity"],
        subset["alcohol"],
        alpha=0.35,
        label=f"Quality {quality_class}"
    )
plt.title("Volatile Acidity vs Alcohol")
plt.xlabel("Volatile Acidity")
plt.ylabel("Alcohol")
plt.legend()
plt.show()
```

### Correlation analysis 

```python
plt.figure(figsize=(10, 8))
correlation_data = data.drop(columns=["quality_original"]).corr(numeric_only=True)
sns.heatmap(
    correlation_data,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()
```

### Data Set loading

```python
RANDOM_STATE = 42
X = data.drop(columns=["quality", "quality_original"])
y = data["quality"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)

print("\nTraining target distribution")
display(y_train.value_counts(normalize=True).round(3))
```
